```python
# Column names were cleaned
df.columns = df.columns.str.strip()
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")

df.columns
```

```python
# student name was fixed with
df["student_name"] =  df["student_name"].str.title().replace("Cyri", "John")

# Duplicate rows were removed
df = df.drop_duplicates(subset=["student_name"])

print(df)


```

```python
# Missing values in Age were replaced with 0
df["age"] = df["age"].fillna(0)

# Missing values in Score were replaced with 0
df["score"] = df["score"].fillna(0)

# Missing values in Student Name were replaced with Unknown
df["student_name"] = df["student_name"].fillna("Unknown")

df
```

```python
 # Outliers in Age were removed
df = df[df["age"] <= 100]

# Outliers in Score were removed
df = df[df["score"] <= 100]

print(df)
```

```python
 # The cleaned dataset was saved
df.to_csv("cleaned_student_data.csv", index=False)
```
